```python
import pandas as pd
dataset=pd.read_csv("Social_Network_Ads.csv")
dataset=pd.get_dummies(dataset,dtype=int,drop_first=True)
dataset=dataset.drop("User ID",axis=1)
ind=dataset[['Age', 'EstimatedSalary', 'Gender_Male']]
dep=dataset[['Purchased']]
```

```python
from sklearn.model_selection import train_test_split
xTrain,xTest,yTrain,yTest=train_test_split(ind,dep,test_size=0.30,random_state=0)
```

```python
#(n_estimators=100, *, criterion='gini', max_depth=None, min_samples_split=2, min_samples_leaf=1, 
#min_weight_fraction_leaf=0.0, max_features='sqrt', max_leaf_nodes=None, min_impurity_decrease=0.0, 
#bootstrap=True, oob_score=False, n_jobs=None, random_state=None, 
#verbose=0, warm_start=False, class_weight=None, ccp_alpha=0.0, max_samples=None, monotonic_cst=None)
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import GridSearchCV
param_grid={'n_estimators':[50,100],'criterion':['gini', 'entropy', 'log_loss'],'max_features':['sqrt', 'log2',100,1000],
          'class_weight':['balanced', 'balanced_subsample','dict']}
grid=GridSearchCV(RandomForestClassifier(),param_grid,refit=True,verbose=3,n_jobs=-1,scoring='f1_weighted')
grid.fit(xTrain,yTrain)
```

```python
gPredict=grid.predict(xTest)
```

```python
print("Best score",grid.best_score_)
print("Best Param",grid.best_params_)
```

```python
from sklearn.metrics import confusion_matrix
cm= confusion_matrix(yTest,gPredict)
cm
```

```python
from sklearn.metrics import classification_report
cr=classification_report(yTest,gPredict)
```

```python
print(cr)
```

```python
from sklearn.metrics import roc_auc_score
yProb=grid.predict_proba(xTest)[:,1]
roc_auc_score(yTest,yProb)
```

```python
from sklearn.metrics import RocCurveDisplay
RocCurveDisplay.from_predictions(yTest,yProb)
```
