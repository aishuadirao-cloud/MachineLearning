```python
import pandas as pd
dataset=pd.read_csv("Social_Network_Ads.csv")
dataset=pd.get_dummies(dataset,dtype='int',drop_first=True)
dataset=dataset.drop('User ID',axis=1)
ind=dataset[['Age', 'EstimatedSalary',  'Gender_Male']]
dep=dataset[['Purchased']]
```

```python
from sklearn.model_selection import train_test_split
xTrain,xTest,yTrain,yTest=train_test_split(ind,dep,test_size=0.30,random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
xTrain=sc.fit_transform(xTrain)
xTest=sc.transform(xTest)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV
#penalty='deprecated', *, C=1.0, l1_ratio=0.0, dual=False, tol=0.0001, fit_intercept=True, intercept_scaling=1, 
#class_weight=None, random_state=None, solver='lbfgs', max_iter=100, verbose=0, warm_start=False, n_jobs=None
param_grid = {'penalty':['l1', 'l2', 'elasticnet', None],'C':[1,100,1000],
               'class_weight':['dict','balanced',None],
              'solver':['lbfgs', 'liblinear', 'newton-cg', 'newton-cholesky', 'sag', 'saga']
            }
grid=GridSearchCV(LogisticRegression(),param_grid,refit=True,verbose=3,n_jobs=-1,scoring='f1_weighted')
grid.fit(xTrain,yTrain)
```

```python
print("Best Score",grid.best_score_)
print("Best Parameters",grid.best_params_)
```

```python
gPredict=grid.predict(xTest)
```

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(yTest,gPredict)
cm
```

```python
from sklearn.metrics import classification_report
cr=classification_report(yTest,gPredict)
print(cr)
```

```python
from sklearn.metrics import roc_auc_score
roc_auc_score(yTest,grid.predict_proba(xTest)[:,1])
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import RocCurveDisplay

yProbs = grid.predict_proba(xTest)[:, 1]

RocCurveDisplay.from_predictions(yTest, yProbs)
plt.plot([0, 1], [0, 1], color='red', linestyle='--')
plt.show()
```
