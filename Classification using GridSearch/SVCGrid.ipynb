```python
import pandas as pd
dataset=pd.read_csv("Social_Network_Ads.csv")
dataset=pd.get_dummies(dataset,dtype='int',drop_first=True)
dataset=dataset.drop('User ID',axis=1)
ind=dataset[['Age', 'EstimatedSalary','Gender_Male']]
dep=dataset[['Purchased']]
```

```python
from sklearn.model_selection import train_test_split
xTrain,xTest,yTrain,yTest=train_test_split(ind,dep,test_size=0.30, random_state=0)
```

```python
from sklearn.preprocessing import StandardScaler
sc=StandardScaler()
xTrain=sc.fit_transform(xTrain)
xTest=sc.transform(xTest)
```

```python
#(*, C=1.0, kernel='rbf', degree=3, gamma='scale', coef0=0.0, shrinking=True, probability='deprecated', tol=0.001, cache_size=200, class_weight=None, 
#verbose=False, max_iter=-1, decision_function_shape='ovr', break_ties=False, random_state=None)
from sklearn.svm import SVC
from sklearn.model_selection import GridSearchCV
param_grid={'kernel':['linear', 'poly', 'rbf', 'sigmoid'],'gamma':['scale', 'auto'],'class_weight':['dict','balanced',None],'probability':[True]}
grid=GridSearchCV(SVC(),param_grid,refit=True,verbose=3,n_jobs=-1,scoring='f1_weighted')
grid.fit(xTrain,yTrain)                  
```

```python
print("Best Score",grid.best_score_)
print("Best Parameters",grid.best_params_)
```

```python
gPredict=grid.predict(xTest)
```

```python
from sklearn.metrics import confusion_matrix
cm=confusion_matrix(yTest,gPredict)
cm
```

```python
from sklearn.metrics import classification_report
cr = classification_report(yTest,gPredict)
print(cr)
```

```python
from sklearn.metrics import roc_auc_score
yProbs = grid.predict_proba(xTest)[:, 1]
roc_auc_score(yTest,yProbs)
```

```python
import matplotlib.pyplot as plt
from sklearn.metrics import RocCurveDisplay
RocCurveDisplay.from_predictions(yTest, yProbs)
```
