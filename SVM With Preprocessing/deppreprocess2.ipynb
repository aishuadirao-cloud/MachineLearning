```python
import pickle
with open('scaler.pkl', 'rb') as f:
    sc = pickle.load(f)
with open('scaler1.pkl', 'rb') as f:
    scy = pickle.load(f)
age=float(input("Age="))
bmi=float(input("BMI="))
noOfChildren=int(input("No of Children="))
sexMale=int(input("Male 1 else 0="))
smoker=int(input("If Smoker 1 else 0"))
```

```python
preinput=sc.transform([[age,bmi,noOfChildren,sexMale,smoker]])
loadedModel=pickle.load(open('SVMFinalgo.sav','rb'))
result=loadedModel.predict(preinput)
result
```

```python
preOutput=scy.inverse_transform([result])
preOutput
```
