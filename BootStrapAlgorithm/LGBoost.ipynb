```python
import lightgbm as lgb
```

```python
from lightgbm import LGBMRegressor
```

```python
import pandas as pd
dataset=pd.read_csv("insurance_pre.csv")
dataset
```

```python
dataset=pd.get_dummies(dataset,dtype=int,drop_first=True)
dataset
```

```python
independentVariable=dataset[['age', 'bmi', 'children','sex_male', 'smoker_yes']]
dependentVariable=dataset[['charges']]
```

```python
from sklearn.model_selection import train_test_split
xTrain,xTest,yTrain,yTest=train_test_split(independentVariable,dependentVariable,test_size=0.30,random_state=0)
```

```python
params = {
    'objective': 'regression',
    'metric': 'rmse',
    'boosting_type': 'gbdt',
    'num_leaves': 31,
    'learning_rate': 0.05,
    'feature_fraction': 0.9
}
```

```python
train_data = lgb.Dataset(xTrain, label=yTrain)
```

```python
regressor=lgb.train(params,train_data)
```

```python
yPredict=regressor.predict(xTest)
from sklearn.metrics import r2_score
rScore=r2_score(yTest,yPredict)
rScore
```
