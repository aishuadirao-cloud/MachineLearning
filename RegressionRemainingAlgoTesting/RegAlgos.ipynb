```python
import pandas as pd
dataset=pd.read_csv("50_Startups.csv")
```

```python
dataset=pd.get_dummies(dataset,dtype='int',drop_first=True)
```

```python
dataset.columns
```

```python
ind=dataset[['R&D Spend', 'Administration', 'Marketing Spend','State_Florida', 'State_New York']]
```

```python
dep=dataset[['Profit']]
```

```python
from sklearn.model_selection import train_test_split
xTrain,xTest,yTrain,yTest=train_test_split(ind,dep,test_size=1/3,random_state=0)
```

```python
from sklearn.linear_model import Ridge
regressor=Ridge()
regressor.fit(xTrain,yTrain)
yPredict=regressor.predict(xTest)
from sklearn.metrics import r2_score
rscore=r2_score(yTest,yPredict)
rscore
```

```python
from sklearn.linear_model import LassoLars
regressor=LassoLars()
regressor.fit(xTrain,yTrain)
yPredict=regressor.predict(xTest)
from sklearn.metrics import r2_score
rscore=r2_score(yTest,yPredict)
rscore
```
